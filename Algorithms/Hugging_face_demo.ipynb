**token_type_ids**<br>
These IDs are used to distinguish between different sequences in tasks that involve multiple sentences, such as question-answering and sentence-pair classification. BERT uses this mechanism to understand which tokens belong to which segment. For single-sequence tasks like sentiment analysis, token_type_ids are all zeros.

**attention_mask** <br>
The attention mask is used to differentiate between actual tokens and padding tokens (if any). It helps the model focus on non-padding tokens and ignore padding tokens. A value of 1 indicates that the token should be attended to, while a value of 0 indicates padding.

**Why Padding Tokens Are Used**<br>
* Uniform Sequence Length: Deep learning models typically process input data in batches. To efficiently process these batches, all sequences in a batch must have the same length. Padding tokens ensure this by extending shorter sequences to match the length of the longest sequence in the batch.
* Efficient Computation: Fixed-length sequences allow for more efficient use of hardware resources, as the model can process all sequences in parallel without needing to handle variable-length sequences individually.

```python
from datasets import load_dataset
dataset = load_dataset('imdb')
```

the model used for the tokenizer must also be used for the model training as well.

```python
#dataset metadata
dataset
```

###  Preprocessing of Data

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

def tokenize_function(examples):
    return tokenizer(examples['text'],padding = "max_length",truncation = True)

tokenize_datasets = dataset.map(tokenize_function,batched = True)
```

```python
tokenize_datasets
```

```python
tokenize_datasets["train"][0]
```

### Setup training arguments

### setup training arguments
specify hyperparameters and training settings

```python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir='./results',
    eval_strategy = "epoch",
    learning_rate = 2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs = 3,
    weight_decay = 0.01
)
training_args
```

## Initializing the model

```python
from transformers import AutoModelForSequenceClassification, Trainer

model = AutoModelForSequenceClassification.from_pretrained('bert-base-uncased',num_labels = 2)

trainer = Trainer(
    model = model,
    args = training_args,
    train_dataset = tokenize_datasets['train'],
    eval_dataset = tokenize_datasets['test']
)
```

### Train the model

```python
trainer.train()
```

### Evaluate the model
Assess model's performance on validation set

```python
results = trainer.evaluate()
print(results)
```

### 💾 What a Checkpoint Contains

A checkpoint file (often stored as a .pth or .ckpt file) typically contains several critical pieces of information:

* Model Weights (Parameters): This is the core of the checkpoint—the actual numerical values (tensors) learned by the model (e.g., the weights and biases of all layers).

* Optimizer State: This includes the internal state variables of the optimization algorithm (like Adam or SGD). For example, it stores the moving averages used by Adam, which are necessary to continue the optimization exactly where it left off.

* Training Epoch/Step: The number of completed epochs or global steps, allowing the training loop to pick up precisely from the next iteration.

* Learning Rate Scheduler State: If you're using a schedule to adjust the learning rate, the checkpoint saves its current state.

* Metrics: Often includes the best-performing metrics achieved so far (e.g., lowest validation loss or highest accuracy), which helps in deciding when to save the next best model.

it helps with early stopping by saving the best performing epoch and restart the training job from previously saved state.

```python
import arxiv
import pandas as pd
```

```python
#query to search for ai related papers
query = "ai OR artificial intelligence OR machine learning"
search = arxiv.Search(query = query,max_results = 10,sort_by = arxiv.SortCriterion.SubmittedDate)

#fetch papers
papers = []
for result in search.results():
    papers.append({
        'published':result.published,
        'title' : result.title,
        'abstract' : result.summary,
        'categories' : result.categories
    })

df = pd.DataFrame(papers)

pd.set_option('display.max_colwidth',None)
df.head(10)
```

```python
#If I want to summarize the research papers,
from transformers import pipeline
abstract = df['abstract'][0]

summarizer = pipeline("summarization", model = "facebook/bart-large-cnn",truncation=True)
summarization_result = summarizer(abstract)
```
